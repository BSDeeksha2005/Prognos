# FD001 exploratory data analysis
Load trajectories, inspect per-engine lifetimes and sensor patterns, then identify near-constant sensors using training variance only. A sensor is omitted only when the training variance is at or below the explicit 1e-8 threshold; retain this rationale and compare plots before changing it.

```python
from pathlib import Path

import matplotlib.pyplot as plt

from rul.data import load_cycles

data = load_cycles(Path("../data/raw/train_FD001.txt"))
sensors = [f"sensor_{i}" for i in range(1, 22)]
print(data.groupby("unit_id").cycle.max().describe())
variance = data[sensors].var()
constant = variance[variance <= 1e-8].index.tolist()
print("Near-constant sensors (training variance <= 1e-8):", constant)
```

```python
fig, axes = plt.subplots(3, 1, figsize=(12, 10))
for axis, sensor in zip(axes, ["sensor_2", "sensor_3", "sensor_14"], strict=False):
    for unit_id, engine in data.groupby("unit_id"):
        if unit_id <= 10:
            axis.plot(engine.cycle, engine[sensor], alpha=0.45, linewidth=1)
    axis.set(title=sensor, xlabel="Cycle", ylabel="Reading")
plt.tight_layout()
plt.show()
```
